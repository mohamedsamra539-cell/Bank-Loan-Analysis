Outlier Handling

After examining the numerical columns (`annual_income`, `loan_amount`, `interest_rate`, `installment`, `debt_to_income_ratio`, `total_payment`, and `total_accounts`) using descriptive statistics and boxplots, no outliers were removed.

This decision was made because the detected extreme values appear to represent **valid customer records rather than data entry errors**. In banking datasets, customers can naturally have very high incomes, large loan amounts, or high total payments. Removing these observations could eliminate important business cases and introduce bias into the analysis.

Only invalid or impossible values (e.g., negative income, negative loan amount, or unrealistic interest rates) would be considered for removal. Since no such values were identified, all records were retained for the analysis.

## cleaning data

```python
Bank["emp_title"]=Bank['emp_title'].fillna("Unknown")
```

```python
date_cols = [
    "issue_date",
    "last_credit_pull_date",
    "last_payment_date",
    "next_payment_date"
]

for col in date_cols:
    Bank[col] = pd.to_datetime(
        Bank[col],
        format="%d-%m-%Y",
        errors="coerce"
    )
```

```python
Bank["emp_length"].value_counts()
```

```python
Bank["emp_length"] = (
    Bank["emp_length"]
    .replace("< 1 year", "1")
    .replace("10+ years", "10")
    .astype(int)
)
```

```python
Bank["emp_length"].value_counts()
```

```python
Bank.rename(columns={"term": "loan_term_months"}, inplace=True)
```
